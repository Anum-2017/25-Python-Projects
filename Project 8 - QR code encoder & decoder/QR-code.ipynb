```python
import qrcode
import cv2
from IPython.display import display, Markdown

# Encoder function
def encode_qr_code(data, output_path):
    qr = qrcode.QRCode(
        version = 1,
        box_size = 10,
        border = 5
    )

    qr.add_data(data)
    qr.make(fit=True)

    img = qr.make_image(fill_color="purple", back_color="white")
    img.save(output_path)

    display(Markdown(f"**QR Code generated and saved successfully at :** \n\n {output_path}"))

# Decoder function
def decode_qr_code(image_path):
    img = cv2.imread(image_path)
    detector = cv2.QRCodeDetector()
    
    data, vertices, _ = detector.detectAndDecode(img)
    
    if data:
        display(Markdown(f"**Decoded QR Code Data :** \n\n {data}"))
    else:
        print("No QR code found in the image.")

qr_data = "www.linkedin.com/in/anum-k-442b2022b"  
output_path = "C:/Users/anumk/Desktop/25 Python Projects/Project 8 - QR code encoder & decoder/QR-code.png"

# Encode QR Code 
encode_qr_code(qr_data , output_path)

# Decode the generated QR Code 
decode_qr_code(output_path)
```
